## Setup: Load Data and Libraries

```
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization defaults
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

# Load the dataset
df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')

print('✅ Data loaded successfully!')
print(f'Dataset: {df.shape[0]} rows × {df.shape[1]} columns')
```

# 📝 Exercise 1 Solutions: Initial Data Exploration

## Question 1.1: Display the first 10 rows

### Solution:

```
# Simple approach
df.head(10)
```

**💡 Explanation:**
- `.head()` shows the first rows (default is 5)
- `.head(10)` specifies exactly 10 rows
- Alternatively, you could use `.head(n=10)`

**🎓 Faculty Tip:** This is your "first glance" at the data - like opening an Excel file

## Question 1.2: What are the column names?

### Solution Approach 1 (Simple):

```
# Approach 1: View as Index object
df.columns
```

### Solution Approach 2 (As List):

```
# Approach 2: Convert to list for easier viewing
print(df.columns.tolist())
print(f"\nTotal columns: {len(df.columns)}")
```

**💡 Explanation:**
- `.columns` returns the column names as an Index object
- `.tolist()` converts it to a regular Python list
- Both approaches are valid - choose based on what you need!

**🎓 Faculty Tip:** Knowing your variable names is step 1 in any analysis

## Question 1.3: Get summary statistics

### Solution:

```
# Get statistical summary of numerical columns
df.describe()
```

### Alternative: Include ALL columns (including categorical):

```
# Include non-numerical columns too
df.describe(include='all')
```

**💡 Explanation:**
- `.describe()` gives you count, mean, std, min, 25%, 50%, 75%, max
- By default, only numerical columns are included
- `include='all'` shows stats for categorical columns too (count, unique, top, freq)

**🎓 Faculty Tip:** This replaces hours of manual Excel calculations!

## Question 1.4: Check for missing data

### Solution Approach 1 (Count):

```
# Count missing values per column
print("Missing values per column:")
print(df.isnull().sum())
```

### Solution Approach 2 (Percentage):

```
# Calculate percentage of missing values
print("Missing data percentage:")
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
print(missing_pct[missing_pct > 0])  # Show only columns with missing data
```

### Solution Approach 3 (Visual):

```
# Visualize missing data
plt.figure(figsize=(12, 6))
sns.heatmap(df.isnull(), yticklabels=False, cbar=True, cmap='viridis')
plt.title('Missing Data Visualization (Yellow = Missing)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**💡 Explanation:**
- `.isnull()` creates a Boolean mask (True/False) for missing values
- `.sum()` counts the True values (missing)
- Visualizations help spot patterns in missing data

**🎓 Faculty Tip:** Always check missing data before analysis - it can bias your results!

# 📊 Exercise 2 Solutions: Answering Research Questions

## Question 2.1: Overall survival rate

### Solution Approach 1 (Simple):

```
# Calculate survival rate
survival_rate = df['Survived'].mean() * 100
print(f"Overall Survival Rate: {survival_rate:.2f}%")
```

### Solution Approach 2 (With More Context):

```
# More detailed survival analysis
survived = df['Survived'].sum()
total = len(df)
survival_rate = (survived / total) * 100

print(f"Survivors: {survived}")
print(f"Total Passengers: {total}")
print(f"Survival Rate: {survival_rate:.2f}%")
print(f"Death Rate: {100 - survival_rate:.2f}%")
```

**💡 Explanation:**
- `.mean()` on a binary (0/1) column gives the proportion
- Multiply by 100 to convert to percentage
- Both approaches work - choose based on how much detail you need

**🎓 Research Insight:** About 38% survived - this is a key baseline metric

## Question 2.2: Average fare by class

### Solution Approach 1 (Basic):

```
# Group by class and calculate mean fare
fare_by_class = df.groupby('Pclass')['Fare'].mean()
print(fare_by_class)
```

### Solution Approach 2 (Formatted):

```
# More readable output
print("Average Fare by Passenger Class:")
print("=" * 40)
for pclass, fare in df.groupby('Pclass')['Fare'].mean().items():
    print(f"Class {pclass}: ${fare:.2f}")
```

### Solution Approach 3 (With Visualization):

```
# Create bar chart
fare_by_class = df.groupby('Pclass')['Fare'].mean()
plt.figure(figsize=(10, 6))
fare_by_class.plot(kind='bar', color=['gold', 'silver', 'bronze'])
plt.title('Average Fare by Passenger Class', fontsize=14, fontweight='bold')
plt.xlabel('Passenger Class')
plt.ylabel('Average Fare ($)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print(fare_by_class.round(2))
```

**💡 Explanation:**
- `.groupby()` splits data into groups
- `['Fare'].mean()` calculates average for each group
- This is like a pivot table in Excel

**🎓 Research Insight:** First class paid ~3x more than third class

## Question 2.3: Gender analysis

### Solution:

```
# Survival rate by gender
survival_by_gender = df.groupby('Sex')['Survived'].mean() * 100
print("Survival Rate by Gender:")
print(survival_by_gender.round(2))

# Also show counts
print("\nPassenger Count by Gender:")
print(df['Sex'].value_counts())
```

### With Visualization:

```
# Create side-by-side comparison
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Survival rate
survival_by_gender.plot(kind='bar', ax=axes[0], color=['lightblue', 'lightcoral'])
axes[0].set_title('Survival Rate by Gender', fontweight='bold')
axes[0].set_ylabel('Survival Rate (%)')
axes[0].set_xlabel('Gender')
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=0)

# Plot 2: Count
df['Sex'].value_counts().plot(kind='bar', ax=axes[1], color=['lightblue', 'lightcoral'])
axes[1].set_title('Passenger Count by Gender', fontweight='bold')
axes[1].set_ylabel('Count')
axes[1].set_xlabel('Gender')
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=0)

plt.tight_layout()
plt.show()
```

**💡 Explanation:**
- Same groupby pattern as before
- Multiply by 100 for percentage
- `.value_counts()` shows distribution

**🎓 Research Insight:** Dramatic difference - "women and children first" policy was real!

## Question 2.4: Age distribution

### Solution:

```
# Age statistics
print("Age Distribution:")
print(f"Minimum Age: {df['Age'].min():.1f} years")
print(f"Maximum Age: {df['Age'].max():.1f} years")
print(f"Average Age: {df['Age'].mean():.1f} years")
print(f"Median Age: {df['Age'].median():.1f} years")
print(f"Std Deviation: {df['Age'].std():.1f} years")
```

### With Visualization:

```
# Create histogram
plt.figure(figsize=(12, 6))
df['Age'].plot(kind='hist', bins=30, edgecolor='black', alpha=0.7)
plt.axvline(df['Age'].mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {df["Age"].mean():.1f}')
plt.axvline(df['Age'].median(), color='green', linestyle='--', linewidth=2, label=f'Median: {df["Age"].median():.1f}')
plt.title('Age Distribution of Passengers', fontsize=14, fontweight='bold')
plt.xlabel('Age (years)')
plt.ylabel('Frequency')
plt.legend()
plt.tight_layout()
plt.show()
```

# 🔍 Exercise 3 Solutions: Data Filtering

## Question 3.1: High fare passengers

### Solution:

```
# Filter for high-fare passengers
high_fare = df[df['Fare'] > 100]

print(f'Number of high-fare passengers: {len(high_fare)}')
print(f'Average fare: ${high_fare["Fare"].mean():.2f}')
print(f'Highest fare: ${high_fare["Fare"].max():.2f}')
print(f'Survival rate: {high_fare["Survived"].mean() * 100:.1f}%')
```

**💡 Explanation:**
- `df[condition]` filters rows where condition is True
- Returns a new DataFrame with matching rows
- You can then analyze this subset

**🎓 Research Tip:** Filtering lets you segment your data for targeted analysis

## Question 3.2: Young survivors

### Solution:

```
# Filter for young survivors (multiple conditions)
young_survivors = df[(df['Age'] < 30) & (df['Survived'] == 1)]

print(f'Young survivors: {len(young_survivors)}')
print(f'Average age: {young_survivors["Age"].mean():.1f} years')
print(f'Average fare: ${young_survivors["Fare"].mean():.2f}')

# Compare to young non-survivors
young_non_survivors = df[(df['Age'] < 30) & (df['Survived'] == 0)]
print(f'\nYoung non-survivors: {len(young_non_survivors)}')
print(f'Average age: {young_non_survivors["Age"].mean():.1f} years')
```

**💡 Explanation:**
- Use `&` for AND (both conditions must be true)
- Use `|` for OR (either condition can be true)
- Parentheses are important: `(condition1) & (condition2)`

**🎓 Common Mistake:** Forgetting parentheses around each condition

## Question 3.3: Class comparison

### Solution:

```
# Compare first and third class
first_class = df[df['Pclass'] == 1]
third_class = df[df['Pclass'] == 3]

print('=' * 50)
print('FIRST CLASS')
print('=' * 50)
print(f'Passengers: {len(first_class)}')
print(f'Average Fare: ${first_class["Fare"].mean():.2f}')
print(f'Survival Rate: {first_class["Survived"].mean()*100:.1f}%')
print(f'Average Age: {first_class["Age"].mean():.1f} years')

print('\n' + '=' * 50)
print('THIRD CLASS')
print('=' * 50)
print(f'Passengers: {len(third_class)}')
print(f'Average Fare: ${third_class["Fare"].mean():.2f}')
print(f'Survival Rate: {third_class["Survived"].mean()*100:.1f}%')
print(f'Average Age: {third_class["Age"].mean():.1f} years')

print('\n' + '=' * 50)
print('COMPARISON')
print('=' * 50)
print(f'Fare Ratio (1st/3rd): {first_class["Fare"].mean() / third_class["Fare"].mean():.2f}x')
print(f'Survival Difference: {(first_class["Survived"].mean() - third_class["Survived"].mean())*100:.1f} percentage points')
```

### With Visualization:

```
# Create comparison plots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Survival comparison
classes = ['First Class', 'Third Class']
survival = [first_class['Survived'].mean() * 100, third_class['Survived'].mean() * 100]
axes[0].bar(classes, survival, color=['gold', 'brown'])
axes[0].set_title('Survival Rate Comparison', fontweight='bold', fontsize=12)
axes[0].set_ylabel('Survival Rate (%)')
axes[0].set_ylim([0, 100])

# Plot 2: Fare comparison
fares = [first_class['Fare'].mean(), third_class['Fare'].mean()]
axes[1].bar(classes, fares, color=['gold', 'brown'])
axes[1].set_title('Average Fare Comparison', fontweight='bold', fontsize=12)
axes[1].set_ylabel('Average Fare ($)')

plt.tight_layout()
plt.show()
```

# 📊 Exercise 4 Solutions: Create Visualizations

## Question 4.1: Fare distribution

### Solution Approach 1 (Basic):

```
# Simple histogram
df['Fare'].plot(kind='hist', bins=30)
plt.title('Fare Distribution')
plt.xlabel('Fare ($)')
plt.ylabel('Frequency')
plt.show()
```

### Solution Approach 2 (Enhanced):

```
# Enhanced histogram with statistics
plt.figure(figsize=(12, 6))
df['Fare'].plot(kind='hist', bins=30, edgecolor='black', alpha=0.7, color='steelblue')
plt.axvline(df['Fare'].mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: ${df["Fare"].mean():.2f}')
plt.axvline(df['Fare'].median(), color='green', linestyle='--', linewidth=2, label=f'Median: ${df["Fare"].median():.2f}')
plt.title('Fare Distribution with Statistics', fontsize=14, fontweight='bold')
plt.xlabel('Fare ($)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Mean fare: ${df['Fare'].mean():.2f}")
print(f"Median fare: ${df['Fare'].median():.2f}")
print(f"Note: Mean > Median suggests right-skewed distribution (some very high fares)")
```

## Question 4.2: Survival by class

### Solution:

```
# Create professional bar plot with Seaborn
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Pclass', y='Survived', palette='Set2', ci=95)
plt.title('Survival Rate by Passenger Class', fontsize=14, fontweight='bold')
plt.xlabel('Passenger Class', fontsize=12)
plt.ylabel('Survival Rate', fontsize=12)
plt.ylim([0, 1])

# Add exact percentages on bars
for i, pclass in enumerate([1, 2, 3]):
    rate = df[df['Pclass'] == pclass]['Survived'].mean()
    plt.text(i, rate + 0.02, f'{rate*100:.1f}%', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```

## Question 4.3: Fare by gender (BONUS)

### Solution:

```
# Box plot comparing fares by gender
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Sex', y='Fare', palette='Set1')
plt.title('Fare Distribution by Gender', fontsize=14, fontweight='bold')
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Fare ($)', fontsize=12)
plt.tight_layout()
plt.show()

print("Average fares by gender:")
print(df.groupby('Sex')['Fare'].mean().round(2))
```

# 🎁 BONUS: Advanced Analysis Examples

## Bonus 1: Create a Comprehensive Dashboard

Combine multiple insights in one view:

```
# Create a 2x2 dashboard
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Comprehensive Passenger Analysis Dashboard', fontsize=18, fontweight='bold', y=0.995)

# Plot 1: Survival by class and gender
survival_pivot = df.groupby(['Pclass', 'Sex'])['Survived'].mean().unstack()
survival_pivot.plot(kind='bar', ax=axes[0, 0], color=['lightcoral', 'lightblue'])
axes[0, 0].set_title('Survival Rate by Class and Gender', fontweight='bold')
axes[0, 0].set_ylabel('Survival Rate')
axes[0, 0].set_xlabel('Class')
axes[0, 0].legend(title='Gender')
axes[0, 0].set_xticklabels(axes[0, 0].get_xticklabels(), rotation=0)

# Plot 2: Age distribution by survival
axes[0, 1].hist([df[df['Survived']==0]['Age'].dropna(), 
                 df[df['Survived']==1]['Age'].dropna()], 
                bins=30, label=['Did not survive', 'Survived'], alpha=0.7)
axes[0, 1].set_title('Age Distribution by Survival', fontweight='bold')
axes[0, 1].set_xlabel('Age')
axes[0, 1].set_ylabel('Frequency')
axes[0, 1].legend()

# Plot 3: Fare vs Age colored by survival
survived = df[df['Survived'] == 1]
died = df[df['Survived'] == 0]
axes[1, 0].scatter(died['Age'], died['Fare'], alpha=0.5, c='red', label='Did not survive', s=20)
axes[1, 0].scatter(survived['Age'], survived['Fare'], alpha=0.5, c='green', label='Survived', s=20)
axes[1, 0].set_title('Age vs Fare by Survival', fontweight='bold')
axes[1, 0].set_xlabel('Age')
axes[1, 0].set_ylabel('Fare ($)')
axes[1, 0].legend()

# Plot 4: Passenger count by embarkation port
embark_counts = df['Embarked'].value_counts()
axes[1, 1].pie(embark_counts.values, labels=embark_counts.index, autopct='%1.1f%%', startangle=90)
axes[1, 1].set_title('Passengers by Embarkation Port', fontweight='bold')

plt.tight_layout()
plt.show()
```

## Bonus 2: Statistical Summary Table

Create a publication-ready summary:

```
# Create comprehensive summary table
summary_data = []

for pclass in [1, 2, 3]:
    class_data = df[df['Pclass'] == pclass]
    summary_data.append({
        'Class': pclass,
        'Count': len(class_data),
        'Survival Rate': f"{class_data['Survived'].mean()*100:.1f}%",
        'Avg Age': f"{class_data['Age'].mean():.1f}",
        'Avg Fare': f"${class_data['Fare'].mean():.2f}",
        'Male %': f"{(class_data['Sex']=='male').mean()*100:.1f}%"
    })

summary_df = pd.DataFrame(summary_data)
print("\n" + "="*70)
print("COMPREHENSIVE CLASS ANALYSIS SUMMARY")
print("="*70)
print(summary_df.to_string(index=False))
print("="*70)
```
