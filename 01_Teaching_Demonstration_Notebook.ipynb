# 📊 Module 2: Data Exploration for Research
## Asking the Right Questions

## 2.1 Understanding Your Variables 🔍

Before any analysis, you need to know what you're working with.

````python
# What columns (variables) do we have?
print("📋 Column Names:")
print(df.columns.tolist())
print(f"\n📊 Total columns: {len(df.columns)}")
````

````python
# What are the data types?
print("🏷️ Data Types:")
print(df.dtypes)
print("\n💡 Tip: 'object' usually means text/categorical data")
````

````python
# How many rows (observations) do we have?
print(f"📈 Total observations: {len(df)} rows")
print(f"📊 Dataset dimensions: {df.shape}")
````

## 2.2 The Missing Data Question ❓

**This is crucial for research integrity!** Missing data can bias your results.

````python
# Check for missing values
print("🔍 Missing Values Count:")
missing = df.isnull().sum()
print(missing[missing > 0])  # Only show columns with missing data

# Calculate percentage missing
print("\n📊 Missing Values Percentage:")
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
print(missing_pct[missing_pct > 0])
````

````python
# Visual representation of missing data
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
sns.heatmap(df.isnull(), yticklabels=False, cbar=True, cmap='viridis')
plt.title('Missing Data Visualization (Yellow = Missing)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("💡 Quick Insight: Look for patterns in missing data - is it random or systematic?")
````

## 2.3 Answering Business Questions with Pandas 📈

Let's answer real research questions using this dataset!

````python
# Question 1: What was the survival rate?
survival_rate = df['Survived'].mean() * 100
print(f"🎯 Overall Survival Rate: {survival_rate:.1f}%")

# Question 2: How many passengers in each class?
print("\n🎫 Passengers by Class:")
print(df['Pclass'].value_counts().sort_index())
````

````python
# Question 3: What's the age distribution?
print("👥 Age Statistics:")
print(df['Age'].describe())

# Question 4: Average fare by class?
print("\n💰 Average Fare by Class:")
fare_by_class = df.groupby('Pclass')['Fare'].mean().round(2)
print(fare_by_class)
````

````python
# Question 5: Gender distribution and survival
print("⚥ Survival Rate by Gender:")
survival_by_gender = df.groupby('Sex')['Survived'].mean() * 100
print(survival_by_gender.round(1))

print("\n⚥ Passenger Count by Gender:")
print(df['Sex'].value_counts())
````

# 🧹 Module 3: Data Cleaning & Transformation
## Making Your Data Analysis-Ready

## 3.1 Filtering Data 🔎

**The Power Move:** Segment your data for targeted analysis!

````python
# Scenario 1: Filter by single condition
# Research Question: "What about first-class passengers?"

first_class = df[df['Pclass'] == 1]
print(f"👑 First Class Passengers: {len(first_class)}")
print(f"   Survival Rate: {first_class['Survived'].mean() * 100:.1f}%")
print(f"   Average Fare: ${first_class['Fare'].mean():.2f}")
````

````python
# Scenario 2: Multiple conditions (AND logic)
# Research Question: "Young first-class passengers?"

young_first_class = df[(df['Pclass'] == 1) & (df['Age'] < 30)]
print(f"👶 Young First Class: {len(young_first_class)} passengers")
print(f"   Survival Rate: {young_first_class['Survived'].mean() * 100:.1f}%")
````

````python
# Scenario 3: Filter by list (OR logic)
# Research Question: "Southampton and Queenstown passengers?"

target_ports = df[df['Embarked'].isin(['S', 'Q'])]
print(f"🚢 Passengers from Southampton or Queenstown: {len(target_ports)}")
print(f"   Survival Rate: {target_ports['Survived'].mean() * 100:.1f}%")
````

````python
# Scenario 4: Text search
# Research Question: "Find all 'Mrs.' passengers"

mrs_passengers = df[df['Name'].str.contains('Mrs', na=False)]
print(f"👩 'Mrs.' Passengers: {len(mrs_passengers)}")
print(f"   Survival Rate: {mrs_passengers['Survived'].mean() * 100:.1f}%")
print(f"   Average Age: {mrs_passengers['Age'].mean():.1f} years")
````

## 3.2 Creating New Variables 🔨

**Transform your data** to create new insights!

````python
# Create a copy to work with (good practice!)
df_enhanced = df.copy()

# 1. Simple calculation - Family size
df_enhanced['FamilySize'] = df_enhanced['SibSp'] + df_enhanced['Parch'] + 1

print("👨‍👩‍👧‍👦 Family Size Distribution:")
print(df_enhanced['FamilySize'].value_counts().sort_index())
````

````python
# 2. Categorize continuous data - Age groups
df_enhanced['AgeGroup'] = pd.cut(df_enhanced['Age'], 
                                   bins=[0, 12, 18, 35, 60, 100],
                                   labels=['Child', 'Teen', 'Adult', 'Middle-Aged', 'Senior'])

print("👥 Passengers by Age Group:")
print(df_enhanced['AgeGroup'].value_counts())
````

````python
# 3. Create boolean (True/False) variable
df_enhanced['IsAlone'] = (df_enhanced['FamilySize'] == 1)

print("🚶 Solo vs Family Travelers:")
print(df_enhanced['IsAlone'].value_counts())

print("\n📊 Survival Rates:")
print(f"   Alone: {df_enhanced[df_enhanced['IsAlone']]['Survived'].mean() * 100:.1f}%")
print(f"   With Family: {df_enhanced[~df_enhanced['IsAlone']]['Survived'].mean() * 100:.1f}%")
````

````python
# 4. Fare per person (for families)
df_enhanced['FarePerPerson'] = df_enhanced['Fare'] / df_enhanced['FamilySize']

print("💰 Fare Per Person Statistics:")
print(df_enhanced['FarePerPerson'].describe().round(2))
````

## 3.3 Handling Missing Data 🩹

**Critical for research integrity!** Different strategies for different situations.

````python
# Check current missing data
print("❓ Missing Values Summary:")
missing_summary = df.isnull().sum()
print(missing_summary[missing_summary > 0])
````

````python
# Strategy 1: Fill missing numerical values with median
# (More robust than mean for skewed data)

df_cleaned = df.copy()
df_cleaned['Age'].fillna(df_cleaned['Age'].median(), inplace=True)

print("✅ Age column filled with median")
print(f"   Missing values in Age: {df_cleaned['Age'].isnull().sum()}")
````

````python
# Strategy 2: Fill categorical data with mode (most common value)
most_common_embarked = df_cleaned['Embarked'].mode()[0]
df_cleaned['Embarked'].fillna(most_common_embarked, inplace=True)

print(f"✅ Embarked filled with most common value: {most_common_embarked}")
print(f"   Missing values in Embarked: {df_cleaned['Embarked'].isnull().sum()}")
````

````python
# Strategy 3: Create 'Unknown' category
df_cleaned['Cabin'].fillna('Unknown', inplace=True)

print("✅ Cabin filled with 'Unknown'")
print(f"   'Unknown' cabins: {(df_cleaned['Cabin'] == 'Unknown').sum()}")
````

````python
# Final check
print("🎉 Final Missing Data Check:")
final_missing = df_cleaned.isnull().sum()
if final_missing.sum() == 0:
    print("   ✅ No missing values! Dataset is clean!")
else:
    print(final_missing[final_missing > 0])
````

# 📊 Module 4: Visualization for Insights
## Turning Data into Stories

## 4.1 Quick Pandas Plots 📈

**Fast and easy** - Perfect for exploratory analysis!

````python
# Bar Chart - Categorical counts
df['Pclass'].value_counts().sort_index().plot(kind='bar', color='steelblue')
plt.title('Passengers by Class', fontsize=14, fontweight='bold')
plt.xlabel('Class')
plt.ylabel('Number of Passengers')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````

````python
# Histogram - Distribution
df['Age'].plot(kind='hist', bins=30, edgecolor='black', color='coral')
plt.title('Age Distribution of Passengers', fontsize=14, fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.axvline(df['Age'].mean(), color='red', linestyle='--', label=f'Mean: {df["Age"].mean():.1f}')
plt.legend()
plt.tight_layout()
plt.show()
````

## 4.2 Beautiful Seaborn Visualizations 🎨

**Publication-ready plots** - Impress your reviewers!

````python
# Visualization 1: Survival Rate by Class and Gender
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Pclass', y='Survived', hue='Sex', ci=95)
plt.title('Survival Rate by Class and Gender', fontsize=16, fontweight='bold')
plt.xlabel('Passenger Class', fontsize=12)
plt.ylabel('Survival Rate', fontsize=12)
plt.legend(title='Gender', fontsize=10)
plt.tight_layout()
plt.show()

print("💡 Insight: Notice the 'women and children first' policy in action!")
````

````python
# Visualization 2: Box Plot - Age distribution by Class
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Pclass', y='Age', palette='Set2')
plt.title('Age Distribution by Passenger Class', fontsize=16, fontweight='bold')
plt.xlabel('Passenger Class', fontsize=12)
plt.ylabel('Age (years)', fontsize=12)
plt.tight_layout()
plt.show()

print("💡 Box plot shows: median (line), quartiles (box), and outliers (dots)")
````

````python
# Visualization 3: Scatter plot with regression line
plt.figure(figsize=(10, 6))
sns.regplot(data=df, x='Age', y='Fare', scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
plt.title('Relationship: Age vs Fare Paid', fontsize=16, fontweight='bold')
plt.xlabel('Age (years)', fontsize=12)
plt.ylabel('Fare (£)', fontsize=12)
plt.tight_layout()
plt.show()

correlation = df[['Age', 'Fare']].corr().iloc[0,1]
print(f"📊 Correlation coefficient: {correlation:.3f}")
````

````python
# Visualization 4: Heatmap - Correlation matrix
# Select only numerical columns
numerical_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
correlation_matrix = df[numerical_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', 
            center=0, square=True, linewidths=1)
plt.title('Correlation Matrix: Key Variables', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

print("💡 Red = positive correlation, Blue = negative correlation")
print("   Numbers closer to 1 or -1 = stronger relationships")
````

````python
# Visualization 5: Violin Plot - Distribution comparison
plt.figure(figsize=(12, 6))
sns.violinplot(data=df, x='Pclass', y='Age', hue='Survived', split=True, palette='muted')
plt.title('Age Distribution: Survivors vs Non-Survivors by Class', fontsize=16, fontweight='bold')
plt.xlabel('Passenger Class', fontsize=12)
plt.ylabel('Age (years)', fontsize=12)
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.tight_layout()
plt.show()

print("💡 Violin plots show the full distribution shape - wider = more passengers")
````

## 4.3 Creating a Research Dashboard 📋

**Pro Tip:** Combine multiple plots for comprehensive analysis!

````python
# Create a 2x2 dashboard
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle('Titanic Dataset: Comprehensive Analysis Dashboard', fontsize=18, fontweight='bold', y=1.00)

# Plot 1: Survival rate by class
survival_by_class = df.groupby('Pclass')['Survived'].mean()
axes[0, 0].bar(survival_by_class.index, survival_by_class.values, color=['gold', 'silver', 'bronze'])
axes[0, 0].set_title('Survival Rate by Class', fontweight='bold')
axes[0, 0].set_xlabel('Class')
axes[0, 0].set_ylabel('Survival Rate')
axes[0, 0].set_ylim([0, 1])

# Plot 2: Age distribution
axes[0, 1].hist(df['Age'].dropna(), bins=30, edgecolor='black', color='skyblue')
axes[0, 1].axvline(df['Age'].mean(), color='red', linestyle='--', linewidth=2, label='Mean')
axes[0, 1].set_title('Age Distribution', fontweight='bold')
axes[0, 1].set_xlabel('Age')
axes[0, 1].set_ylabel('Frequency')
axes[0, 1].legend()

# Plot 3: Gender survival comparison
gender_survival = df.groupby('Sex')['Survived'].mean()
axes[1, 0].bar(gender_survival.index, gender_survival.values, color=['lightcoral', 'lightblue'])
axes[1, 0].set_title('Survival Rate by Gender', fontweight='bold')
axes[1, 0].set_xlabel('Gender')
axes[1, 0].set_ylabel('Survival Rate')
axes[1, 0].set_ylim([0, 1])

# Plot 4: Fare distribution by class
df.boxplot(column='Fare', by='Pclass', ax=axes[1, 1])
axes[1, 1].set_title('Fare Distribution by Class', fontweight='bold')
axes[1, 1].set_xlabel('Class')
axes[1, 1].set_ylabel('Fare (£)')
plt.suptitle('')  # Remove default title

plt.tight_layout()
plt.show()

print("🎯 Dashboard created! Perfect for presentations and papers.")
````
